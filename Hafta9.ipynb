```python
uzaylılar = []
```

```python
for uzaylı_sayısı in range(30):
    yeni_uzaylı = {"renk": "yeşil",
                   "puan": 5,
                   "hız": "yavaş"}
    uzaylılar.append(yeni_uzaylı)
```

```python
for uzaylı in uzaylılar[:5]:
    if uzaylı["renk"] == "yeşil":
        uzaylı["renk"] = "sarı"
        uzaylı["hız"] = "orta"
        uzaylı["puan"] = 10
```

```python
for uzaylı in uzaylılar[:10]:
    if uzaylı["renk"] == "sarı":
        uzaylı["renk"] = "kırmızı"
        uzaylı["hız"] = "hızlı"
        uzaylı["puan"] = 15
    if uzaylı["renk"] == "yeşil":
        uzaylı["renk"] = "sarı"
        uzaylı["hız"] = "orta"
        uzaylı["puan"] = 10
for uzaylı in uzaylılar[:15]:
    print(uzaylı)       
```

```python
pizza = {"kenar": "kalın",
        "malzemeler": ["sucuk", "mantar", "biber", "peynir"]}

print(f"Aşağıdaki malzemelere sahip bir {pizza['kenar']} kenarlı pizza sipariş ettiniz:")
for malzeme in pizza["malzemeler"]:
    print(f"- {malzeme}")
```

```python
favori_diller = {
    "ali": ["python", "c++"],
    "veli": ["java", "c#"],
    "ayşe": ["ruby", "javascript"],
    "fatma": ["go", "swift", "java"]
}

for isim, diller in favori_diller.items():
    print(f"{isim.title()}'in favori programlama dilleri şunlardır:")
    for dil in diller:
        print(f"- {dil}")
```

```python
favori_diller = {
    "ali": ["python", "c++"],
    "veli": ["java", "c#"],
    "ayşe": ["ruby", "javascript"],
    "fatma": ["go", "swift", "java"]
}

for isim, diller in favori_diller.items():
    print(f"{isim.title()}'in favori programlama dilleri şunlardır:")
    for dil in diller:
        for harf in dil:
            print(f"- {harf}")
```

```python
kullanıcılar = {
    "aeinstein": {
        "ad": "albert",
        "soyad": "einstein",
        "doğum_yılı": 1879,
        "şehir": "princeton"
    },
    "mcurie": {
        "ad": "marie",
        "soyad": "curie",
        "doğum_yılı": 1867,
        "şehir": "paris"
    }}

for kullanıcı_adı, bilgi in kullanıcılar.items():
    print(f"Kullanıcı Adı: {kullanıcı_adı}")
    tam_ad = f"{bilgi['ad'].title()} {bilgi['soyad'].title()}"
    doğum_yılı = bilgi['doğum_yılı']
    şehir = bilgi['şehir'].title()
    print(f"\tTam Ad: {tam_ad}"
          f"\n\tDoğum Yılı: {doğum_yılı}"
          f"\n\tŞehir: {şehir}")
```
