```python
class Araba:
    def __init__(self, marka, model, yil):
        self.marka = marka
        self.model = model
        self.yil = yil
        self.km = 0

    def bilgileri_goster(self):
        uzun_format = f"{self.marka} {self.model} {self.yil} ({self.km} km)"
        return uzun_format
    
    def km_guncelle(self,km):
        if km >= self.km:
            self.km = km
        else:
            print("Hata: Kilometre geri alınamaz.")

    def km_ekle(self, km):
        if km >= 0:
            self.km += km
        else:
            print("Hata: Eklenen kilometre pozifi olmalıdır.")

    def benzin_doldur(self, litre):
        print(f"{litre} litre benzin dolduruldu.")
```

```python
class ElektrikliAraba(Araba):
    def __init__(self, marka, model, yil, batarya_kapasitesi):
        super().__init__(marka, model, yil)
        self.batarya_kapasitesi = batarya_kapasitesi
        self.batarya_kapasitesi = batarya_kapasitesi

    def batarya_bilgisi(self):
        print(f"{self.marka} {self.model} batarya kapasitesi: {self.batarya_kapasitesi} kWh")
```

```python
ilk_elektrikli_arabam = ElektrikliAraba("Tesla", "Model 3", 2021, 75)
ilk_elektrikli_arabam.batarya_bilgisi()

ilk_arabam = Araba("Toyota", "Corolla", 2020)
ilk_arabam.batarya_bilgisi()
```

```python
ilk_arabam.benzin_doldur(50)
ilk_elektrikli_arabam.benzin_doldur(50)
```

```python
class Batarya:
    def __init__(self, kapasite_kwh):
        self.kapasite_kwh = kapasite_kwh

    def bilgileri_goster(self):
        print(f"Batarya kapasitesi: {self.kapasite_kwh} kWh")

    def mesafe_hesapla(self):
        if self.kapasite_kwh == 75:
            mesafe = 350
            print(f"Bu batarya ile tahmini menzil: {mesafe} km")
        elif self.kapasite_kwh == 100:
            mesafe = 500
            print(f"Bu batarya ile tahmini menzil: {mesafe} km")
        else:
            mesafe = "Bilinmeyen batarya kapasitesi"
            print(mesafe)
```

```python
class ElektrikliAraba(Araba):
    def __init__(self, marka, model, yil, batarya_kapasitesi):
        super().__init__(marka, model, yil)
        self.batarya = Batarya(batarya_kapasitesi)

    def benzin_doldur(self, litre):
        print("Hata: Elektrikli arabalar benzin doldurmaz. ")
```

```python
ikinci_elektrikli_arabam = ElektrikliAraba("Togg", "T10X", 2023, 100)
ikinci_elektrikli_arabam.batarya.bilgileri_goster()
ikinci_elektrikli_arabam.batarya.mesafe_hesapla()
```
