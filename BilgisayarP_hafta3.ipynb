```python
class laptop:
    def __init__(self, marka, model):
        self.marka = marka
        self.model = model
        self.islemci = "Intel Core i7"
        self.ram = "16GB"

    def bilgileri_göster(self):
        return f"Marka: {self.marka}, Model: {self.model}, İşlemci: {self.islemci}, RAM: {self.ram}"
```

```python
class bilgisayar(laptop):
    def __init__(self, marka, model, islemci, ram):
        super().__init__(marka, model)
        self.islemci = islemci
        self.ram = ram
```

```python
ilk_laptopum = laptop("Dell", "XPS 13")
print(ilk_laptopum.islemci)  
print(ilk_laptopum.ram) 

ilk_bilgisayarım = bilgisayar("HP", "Pavilion", "AMD Ryzen 5", "8GB")
print(ilk_bilgisayarım.islemci)  
print(ilk_bilgisayarım.ram)
```
