```python
# RESTORAN SİPARİŞ OTOMASYON SİSTEMİ
# Bu program müşteriden sipariş alır, kategorilere göre ürün sunar ve adisyon keser.
# Her kategori kendi içinde ürün adı ve fiyatı tutar.
corba_menusu = {
    'Mercimek': 40,
    'Ezogelin': 40,
    'Domates': 35,
    'Yayla': 35
}

yemek_menusu = {
    'Adana': 180,
    'Urfa': 170,
    'Tavuk Şiş': 140,
    'Izgara Köfte': 160,
    'Lahmacun': 60
}

icecek_menusu = {
    'Ayran': 15,
    'Kola': 25,
    'Salgam': 15,
    'Su': 5,
    'Cay': 10
}

siparis_listesi = []
toplam_tutar = 0


print("* LEZZET DURAĞI RESTORANINA HOŞ GELDİNİZ*")

program_acik = True

while program_acik:
    print("\n--- ANA MENÜ ---")
    print("1. Çorbalar")
    print("2. Ana Yemekler")
    print("3. İçecekler")
    print("4. Siparişleri Tamamla ve Hesabı Öde")
    print("5. İptal / Çıkış")

    ana_secim = input("Lütfen menüden bir numara seçiniz (1-5): ")

    # --- KATEGORİ 1: ÇORBALAR ---
    if ana_secim == '1':
        print("\n--- ÇORBA MENÜSÜ ---")
        
        for ad, fiyat in corba_menusu.items():
            print(f"- {ad}: {fiyat} TL")
        
        siparis = input("Hangi çorbayı istersiniz? (Ana menüye dönmek için 'x'): ").title()
        
        if siparis in corba_menusu:
            fiyat = corba_menusu[siparis]
            siparis_listesi.append(siparis)
            toplam_tutar = toplam_tutar + fiyat
            print(f">> {siparis} siparişlere eklendi. (+{fiyat} TL)")
        elif siparis == 'X':
            print("Ana menüye dönülüyor...")
        else:
            print("HATA: Böyle bir çorba menüde yok!")

    # --- KATEGORİ 2: ANA YEMEKLER ---
    elif ana_secim == '2':
        print("\n--- ANA YEMEK MENÜSÜ ---")
        for ad, fiyat in yemek_menusu.items():
            print(f"- {ad}: {fiyat} TL")
            
        siparis = input("Hangi yemeği istersiniz? (Ana menüye dönmek için 'x'): ").title()
        
        if siparis in yemek_menusu:
            fiyat = yemek_menusu[siparis]
            siparis_listesi.append(siparis)
            toplam_tutar += fiyat 
            print(f">> {siparis} siparişlere eklendi. (+{fiyat} TL)")
        elif siparis == 'X':
            print("Ana menüye dönülüyor...")
        else:
            print("HATA: Böyle bir yemek menüde yok!")

    # --- KATEGORİ 3: İÇECEKLER ---
    elif ana_secim == '3':
        print("\n--- İÇECEK MENÜSÜ ---")
        for ad, fiyat in icecek_menusu.items():
            print(f"- {ad}: {fiyat} TL")
            
        siparis = input("Ne içmek istersiniz? (Ana menüye dönmek için 'x'): ").title()
        
        if siparis in icecek_menusu:
            fiyat = icecek_menusu[siparis]
            siparis_listesi.append(siparis)
            toplam_tutar += fiyat
            print(f">> {siparis} siparişlere eklendi. (+{fiyat} TL)")
        elif siparis == 'X':
            print("Ana menüye dönülüyor...")
        else:
            print("HATA: Böyle bir içecek menüde yok!")

    # --- HESAP ÖDEME VE FİŞ KESME ---
    elif ana_secim == '4':
        print("\n" + "="*30)
        print("     ADİSYON / FİŞ DETAYI")
        print("="*30)
        
        if len(siparis_listesi) > 0:
            sayac = 1
            # Sipariş listesini yazdırma
            for urun in siparis_listesi:
                print(f"{sayac}. {urun}")
                sayac += 1 # Sayacı artırma
            
            print("-" * 30)
            print(f"TOPLAM TUTAR: {toplam_tutar} TL")
            print("=" * 30)
            print("Bizi tercih ettiğiniz için teşekkürler!")
            
        
            program_acik = False 
        else:
            print("Henüz sipariş vermediniz.")

    # --- ÇIKIŞ ---
    elif ana_secim == '5':
        print("Sipariş verilmeden çıkış yapılıyor...")
        break

    else:
        print("Hatalı bir seçim yaptınız, lütfen tekrar deneyin.")
```
