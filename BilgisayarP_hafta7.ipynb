```python
import json

def kayıtlı_kullanıcılar_al():
    try:
        with open("kullanıcılar.json", "r") as dosya:
            kullanıcılar = json.load(dosya)
    except FileNotFoundError:
        return None
    else:
        return kullanıcılar
    
def yeni_kullanıcı_kayıt():
    kullanıcı_adı = input("Lütfen kullanıcı adınızı girin: ")
    with open("kullanıcılar.json", "w") as dosya:
        json.dump(kullanıcı_adı, dosya)
        print(f"Geri geldiğinizde {kullanıcı_adı} olarak hatırlayacağım.")
    return kullanıcı_adı

def kullanıcı_girişi():
    kullanıcı = input("Lütfen kullanıcı adınızı girin: ")
    kullanıcılar = kayıtlı_kullanıcılar_al()

    if kullanıcı == kullanıcılar:
        print(f"Tekrar hoş geldiniz, {kullanıcılar}!") 
    else:
        onay = input("Bu kullanıcı adı kayıtlı değil. Yeni kullanıcı kaydı oluşturmak ister misiniz? (E/H): ")
        if onay.lower() == "e":
            kullanıcı_adı = yeni_kullanıcı_kayıt()
            print(f"Hoş geldiniz, {kullanıcı_adı}!")
        else:
            print("Kayıt oluşturulmadı. Kullanıcı girişi başarısız oldu.")
```

```python
kullanıcı_girişi()
```
