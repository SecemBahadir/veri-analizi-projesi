```python
from random import choice

class rasgeleYuruyus():
    "Rasgele yürüyüş oluşturmak için bir sınıf"
    def __init__(self, adim_sayisi=5000): 
        self.adim_sayisi = adim_sayisi

        self.x_degeri = [0]
        self.y_degeri = [0]
    def adimlar(self):
        "Bütün yürüyüşü oluştur." 
        while len(self.x_degeri) < self.adim_sayisi:
            x_yon = choice([1, -1])
            x_uzunluk = choice([0, 1, 2, 3, 4])
            x_adim = x_yon * x_uzunluk

            y_yon = choice([1, -1])
            y_uzunluk = choice([0, 1, 2, 3, 4])
            y_adim = y_yon * y_uzunluk

            if x_adim == 0 and y_adim == 0:
                continue

            x = self.x_degeri[-1] + x_adim
            y = self.y_degeri[-1] + y_adim

            self.x_degeri.append(x)
            self.y_degeri.append(y)
```

```python
import matplotlib.pyplot as plt

while True:
    yuruyus = rasgeleYuruyus()
    yuruyus.adimlar()

    fig, ax = plt.subplots(figsize=(10, 6))
    adım_sayisi = range(yuruyus.adim_sayisi)
    ax.scatter(yuruyus.x_degeri, yuruyus.y_degeri, s=15, c=adım_sayisi, cmap=plt.cm.Blues, alpha=0.5)
    ax.scatter(0, 0, c='red', s=100, edgecolors='black')  # Başlangıç noktasını vurgula
    ax.scatter(yuruyus.x_degeri[-1], yuruyus.y_degeri[-1], c='green', s=100, edgecolors='black')  # Bitiş noktasını vurgula
    ax.get_xaxis().set_visible(False)
    ax.get_yaxis().set_visible(False)
    plt.show()
    tekrar = input("Tekrar yürüyüş oluşturmak ister misiniz? (evet/hayır): ")
    if tekrar.lower() != 'evet':
        break
```

```python
from random import randint

class Zar():
    "Zar atma simülasyonu için bir sınıf"
    def __init__(self, yuzey_sayisi=6):
        self.yuzey_sayisi = yuzey_sayisi

    def at(self):
        "Zarı at ve sonucunu döndür."
        return randint(1, self.yuzey_sayisi)
```

```python
import plotly.express as px
zar = Zar()

sonuclar=[]
for zar_atımı in range(1000):
    sonuc = zar.at()
    sonuclar.append(sonuc)

frekanslar = {yuzey: sonuclar.count(yuzey) for yuzey in range(1, zar.yuzey_sayisi + 1)}

fig = px.bar(x=list(frekanslar.keys()), y=list(frekanslar.values()), labels={'x': 'Zar Yüzeyi', 'y': 'Frekans'}, title='Zar Atma Sonuçları')
fig.show()
```
