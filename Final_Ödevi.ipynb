```python
def main():
    # ŞART 3: Liste veri yapısı (Yapılan işlemleri hafızada tutmak için)
    islem_gecmisi = []

    print("--- Akıllı Hesap Makinesi ---")
    print("İşlemler: +, -, *, /")
    print("Ekstra Komutlar: 'h' (Geçmişi Gör), 'q' (Çıkış)")

    # ŞART 1: While Döngüsü (Program sürekli çalışsın)
    while True:
        # ŞART 5: Kullanıcıdan veri alma (String olarak başlar)
        giris = input("\nBirinci sayıyı girin (veya komut 'q'/'h'): ")

        # Çıkış ve Geçmiş kontrolü
        if giris.lower() == 'q':
            print("Hesap makinesi kapatılıyor...")
            break
        
        elif giris.lower() == 'h':
            print("\n--- İŞLEM GEÇMİŞİ ---")
            # ŞART 1 (Ekstra): For döngüsü ile listeyi gezme
            if len(islem_gecmisi) == 0:
                print("Henüz bir işlem yapılmadı.")
            else:
                for kayit in islem_gecmisi:
                    print(kayit)
            continue # Döngünün başına dön, sayı sorma kısmına geri git

        # Eğer komut girilmediyse, hesaplama işlemine devam et
        try:
            # ŞART 5: String'i Float'a çevirme (Ondalıklı sayı olabilir)
            sayi1 = float(giris)
            islem = input("İşlemi seçin (+, -, *, /): ")
            sayi2 = float(input("İkinci sayıyı girin: "))
            
            sonuc = 0
            
            # ŞART 2: If-Elif-Else Koşullu Yapılar
            if islem == '+':
                sonuc = sayi1 + sayi2
            elif islem == '-':
                sonuc = sayi1 - sayi2
            elif islem == '*':
                sonuc = sayi1 * sayi2
            elif islem == '/':
                if sayi2 == 0:
                    print("Hata: Bir sayı 0'a bölünemez!")
                    continue # Döngü başına dön
                sonuc = sayi1 / sayi2
            else:
                print("Geçersiz işlem operatörü!")
                continue

            # Sonucu ekrana yazdır
            print(f"SONUÇ: {sonuc}")

            # ŞART 4: Metot kullanımı (.append)
            # Yapılan işlemi formatlayıp listeye ekliyoruz
            kayit_metni = f"{sayi1} {islem} {sayi2} = {sonuc}"
            islem_gecmisi.append(kayit_metni)

        except ValueError:
            print("Hata: Lütfen geçerli bir sayı giriniz!")

if __name__ == "__main__":
    main()
```
