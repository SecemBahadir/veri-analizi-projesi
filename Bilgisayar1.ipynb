```python
def profil_olustur(ad, soyad, **kullanıcı_adı):
    profil = {}
    profil['ad'] = ad
    profil['soyad'] = soyad
    for anahtar, değer in kullanıcı_adı.items():
        profil[anahtar] = değer
    return profil
    print("Profil oluşturuluyor...")
```

```python
secembahadirprofile=profil_olustur('Secem', 'Bahadir', yaş=21, şehir='İstanbul', meslek='istatistikçi')
print(secembahadirprofile)
```
