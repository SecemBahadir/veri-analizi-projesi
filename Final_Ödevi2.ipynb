```python
# PROJE: ÖĞRENCİ NOT TAKİP SİSTEMİ

# [KURAL 3] Veri depolama yapısı: Liste tanımlama
ogrenci_listesi = []

print("--- Öğrenci Not Sistemine Hoş Geldiniz ---")

# [KURAL 1] While Döngüsü kullanımı (Program 'q' denilene kadar çalışır)
while True:
    print("\nMENÜ:")
    print("1. Öğrenci Ekle")
    print("2. Listeyi Göster")
    print("3. Son Eklenen Öğrenciyi Sil")
    print("4. Çıkış (q)")

    # [KURAL 5] String değişken kullanımı
    secim = input("Seçiminiz (1-4 veya q): ")

    # [KURAL 2] If ile koşullu yapı kullanımı
    if secim == '1':
        # Kullanıcıdan verileri alıyoruz
        # [KURAL 4] Metot kullanımı: .title() (Baş harfleri büyütür) [PDF Kaynak: 3469]
        ad = input("Öğrenci Adı: ").title()
        soyad = input("Öğrenci Soyadı: ").title()
        
        # [KURAL 5] Integer değişken kullanımı [PDF Kaynak: 5237]
        vize = int(input("Vize Notu: "))
        final = int(input("Final Notu: "))

        # Basit not hesaplama
        ortalama = (vize * 0.4) + (final * 0.6)

        # Başarı durumu belirleme
        durum = "KALDI"
        if ortalama >= 50:
            durum = "GEÇTİ"

        # [KURAL 3] Veri depolama yapısı: Sözlük (Dictionary) [PDF Kaynak: 4747]
        # Verileri bir sözlükte paketliyoruz
        yeni_ogrenci = {
            'ad': ad,
            'soyad': soyad,
            'ort': ortalama,
            'durum': durum
        }

        # [KURAL 4] Metot kullanımı: .append() (Listeye ekler) [PDF Kaynak: 3698]
        ogrenci_listesi.append(yeni_ogrenci)
        print(f">> {ad} sisteme eklendi.")

    elif secim == '2':
        print("\n--- SINIF LİSTESİ ---")
        # [KURAL 1] For döngüsü kullanımı (Listeyi gezmek için) [PDF Kaynak: 3906]
        if len(ogrenci_listesi) > 0:
            for ogrenci in ogrenci_listesi:
                # Sözlükten verileri okuma
                print(f"{ogrenci['ad']} {ogrenci['soyad']} -> Ort: {ogrenci['ort']} ({ogrenci['durum']})")
        else:
            print("Listede henüz öğrenci yok.")

    elif secim == '3':
        if len(ogrenci_listesi) > 0:
            # [KURAL 4] Metot kullanımı: .pop() (Son elemanı siler) [PDF Kaynak: 3730]
            silinen = ogrenci_listesi.pop()
            print(f">> {silinen['ad']} listeden silindi.")
        else:
            print("Silinecek öğrenci yok.")

    elif secim == 'q':
        print("Programdan çıkılıyor...")
        break  # Döngüyü kırar ve bitirir

    else:
        print("Hatalı tuşlama yaptınız.")
```
